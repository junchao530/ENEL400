```python
# Constants
vt = 25e-3  # thermal voltage
va = 120    # Early voltage
avo = 120   #No load AC voltage gain
beta = 160
vbe_on = 0.7
vcc = 15

# Function to calculate gm
def get_gm(ic):
    return ic / vt

# Function to calculate ro
def get_ro(ic):
    return va / ic

# Function to calculate rc
def get_rc(gm, ro):
    num = (avo / gm) * ro
    denom = (ro - avo / gm)
    return num / denom

# Function to calculate vce
def get_vce(ic, rc):
    return vcc - ic * rc

def get_ib(ic):
    return ic/beta

def get_ie(ic):
    return ic + get_ib(ic)

def get_re(ic):
    ie = get_ie(ic)
    return vt/ie

def get_rpi(ic):
    return (vt/get_ib(ic))

def get_r1(ib):
    return (vcc-vbe_on)/(3*ib)

def get_r2(ib):
    return vbe_on/(2*ib)
    
def engineering_notation(value):
    exponent = int(f"{value:.0e}".split("e")[1])
    # Adjust exponent to the nearest multiple of 3
    engineering_exp = exponent - (exponent % 3)
    scaled_value = value / (10 ** engineering_exp)
    
    # Map exponents to metric prefixes
    prefix = {
        -9: "n",  # nano
        -6: "µ",  # micro
        -3: "m",  # milli
         0: "",   # no prefix
         3: "k",  # kilo
         6: "M",  # mega
         9: "G"   # giga
    }.get(engineering_exp, f"e{engineering_exp}")  # Default to scientific notation if out of range
    
    return f"{scaled_value:.6g} {prefix}"

# Function to get all values with print statements in organized format
def get_values(ic):
    gm = get_gm(ic)
    ro = get_ro(ic)
    rc = get_rc(gm, ro)
    vce = get_vce(ic, rc)
    ib = get_ib(ic)
    ie = get_ie(ic)
    re = get_re(ic)
    rpi = get_rpi(ic)
    r1 = get_r1(ib)
    r2 = get_r2(ib)
    
    # Print all values organized by categories
    print("\n--- Small Signal Models ---")
    print(f"rpi: {engineering_notation(rpi)}Ω")
    print(f"re: {engineering_notation(re)}Ω")
    print(f"ro: {engineering_notation(ro)}Ω")
    print(f"gm: {engineering_notation(gm)}S")
    
    print("\n--- Resistor Values ---")
    print(f"rc: {engineering_notation(rc)}Ω")
    print(f"r1: {engineering_notation(r1)}Ω")
    print(f"r2: {engineering_notation(r2)}Ω")
    
    print("\n--- Biasing Values ---")
    print(f"ib: {engineering_notation(ib)}A")
    print(f"ic: {engineering_notation(ic)}A")
    print(f"vce: {engineering_notation(vce)}V")

    return gm, ro, rc, vce, ib, ie, re, rpi, r1, r2

# Example usage
ic = 1e-3
get_values(ic)
```
